### LeetCode 3005 — Count Elements With Maximum Frequency

#### Question Type
- Array
- Hash Map
- Frequency Counting

#### Problem
You are given an integer array `nums`.

The frequency of an element is the number of times it appears in the array.

Find the highest frequency among all elements.

Return the total number of elements whose values have that highest frequency.

#### Example 1
**Input:** `nums = [1, 2, 2, 3, 1, 4]`  
**Output:** `4`

**Explanation:**  
`1` appears `2` times.  
`2` appears `2` times.  

The maximum frequency is `2`, so the total is:

`2 + 2 = 4`

#### Example 2
**Input:** `nums = [1, 2, 3, 4, 5]`  
**Output:** `5`

**Explanation:**  
Every number appears once, so every element has the maximum frequency.

````python
class Solution:
    def maxFrequencyElements(self, nums: list[int]) -> int:
        freq = {}

        for num in nums:
            # .get read a value from a dictionary
            # Look for the key num. If it exists, return its value. If it does not exist, return 0.
            freq[num] = freq.get(num, 0) + 1

        max_freq = max(freq.values())
        total = 0

        for value in freq.values():
            if value = max_frequency:
            total += value

        return total 
        
````

Find two lines, which together with x-axis forms a container, such that the container contains the most water.

<small>The above vertical lines are represented by array [1,8,6,2,5,4,8,3,7]. In this case, the max area of water (blue section) the container can containis 49. </small>



<strong>Example:</strong>

<pre>

<strong>Output:</strong> 49</pre>

````python
'''
 Two loops approach:
 Find the max area by make 2 for loops to check every possible area
 Time complexity: O(n^2)
 Space complexity: O(1)
 Use two pointers approach is better
'''

class Solution:
    def maxArea(self, height: list[int]) -> int:
        res = 0
        left, right = 0, len(height)-1
        while left <right:
            area = min(height[left], height[right]) * (left - right) 
            res = max(res, area)
            if height[left] < height[right]:
                left+=1
            else:
                right-=1
        return res
````

word2:    p   q   r
merged: a p b q c r
Example 2:

word1:  a   b 
word2:    p   q   r   s
merged: a p b q   r   s
Example 3:

Input: word1 = "abcd", word2 = "pq"
Output: "apbqcd"
Explanation: Notice that as word1 is longer, "cd" is appended to the end.
word1:  a   b   c   d
word2:    p   q 
merged: a p b q c   d


````python
class Solution:
    def mergeAlternatively(self, word1:str, word2:str)->str:
        res=[]
        i,j = 0,0

        while i< len(word1) and j < len(word2):
            res.append(word1[i])
            res.append(word2[j])
            i+=1
            j+=1

        res.append(word1[i:])
        res.append(word2[j:])
        
        return ''.join(res) 
````

````python
Class Solution:
    def search(nums: list[int], target: int) -> int:
        l,r = 0, len(nums) - 1

        while l <= r:
            m = (l + r) // 2
            if nums[m] > target:
                r = m - 1
            elif nums[m] < target:
                l = m + 1
            else:
                return m
        return -1
        
    
````

Determine whether the input string is valid.

A string is valid if:

1. Every opening bracket has a matching closing bracket.
2. Brackets are closed in the correct order.
3. Every closing bracket matches the most recent unmatched opening bracket.

Examples:

Example 1:
Input: s = "()"
Output: True

Example 2:
Input: s = "()[]{}"
Output: True

Example 3:
Input: s = "(]"
Output: False

Example 4:
Input: s = "([)]"
Output: False

Example 5:
Input: s = "{[]}"
Output: True

Constraints:

1 <= len(s) <= 10,000

s contains only:
'(', ')', '{', '}', '[' and ']'

````python
class Solution:
    def isValid(self, s: str) -> bool:
        stack = []
        
        pairs = {
            ")": "(",
            "]": "[",
            "}": "{"
        }

        for char in s:
            # Opening bracket
            if char in "([{":
                stack.append(char)
            # Closing bracket
            else:
                if not stack:
                    return False
                if pairs[char] != stack[-1]:
                    return False
                # closed latest open bracket, then remove the open racket.
                stack.pop()
            return len(stack) == 0
````

### LeetCode 1047 — Remove All Adjacent Duplicates in String

#### Problem

You are given a string `s` containing lowercase English letters.

Repeatedly remove two adjacent characters when they are the same.

Continue until no more adjacent duplicate characters can be removed.

Return the final string.

#### Example 1

```text
Input:
s = "abbaca"

Output:
"ca"

````python
class Solution:
    def removeDuplicates(self, s: str) -> str:
        stack = []

        for char in s:
            if stack and stack[-1] == char:
                stack.pop()
            else:
                stack.append(char)
            return "",join(stack)
                
````

#### Question Type
- String
- Array
- Two Pointers
- In-Place Modification

#### Problem
Write a function that reverses a string.

The input string is given as an array of characters `s`.

You must reverse the array in place and use only constant extra memory.

#### Example 1
**Input:** `s = ["h", "e", "l", "l", "o"]`  
**Output:** `["o", "l", "l", "e", "h"]`

#### Example 2
**Input:** `s = ["H", "a", "n", "n", "a", "h"]`  
**Output:** `["h", "a", "n", "n", "a", "H"]`

#### Constraints
- `1 <= s.length <= 100000`
- Every element in `s` is a printable character.
- Reverse `s` in place.
- Use only constant extra memory.

#### Goal
Use two pointers:
1. Put one pointer at the beginning.
2. Put one pointer at the end.
3. Swap the two characters.
4. Move both pointers toward the middle.
5. Stop when the pointers meet or cross.

````python
class Solution:
    def reverseString(self, s: list[str]) -> None:
        l = 0
        r = len(s) - 1
        While l < r:
            s[l],s[r] = s[r], s[l]
            l += 1
            r -= 1
````
